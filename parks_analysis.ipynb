```python
%pip install pandas openpyxl
import pandas as pd

# Load the CSV file
df = pd.read_csv("parks_payroll_data.csv")

# Display the first 5 rows
df.head()
```

```python
df.info()
```

```python
# Grouping by department and calculating salary stats
dept_summary = df.groupby('department_name').agg(
    total_payroll=('salary', 'sum'),
    average_salary=('salary', 'mean'),
    employee_count=('employee_id', 'count')
).reset_index()

# Rounding average salary to 2 decimal places
dept_summary['average_salary'] = dept_summary['average_salary'].round(2)

# Displaying the summary table
dept_summary
```

```python
# Create a copy of the dataframe to clean
df_clean = df.copy()

# Fill missing department names with "Unassigned"
df_clean['department_name'] = df_clean['department_name'].fillna('Unassigned')

# Export both the detailed cleaned data AND the department summary to Excel
with pd.ExcelWriter('parks_final_report.xlsx') as writer:
    df_clean.to_excel(writer, sheet_name='Employee_Data', index=False)
    dept_summary.to_excel(writer, sheet_name='Dept_Summary', index=False)

print("Excel report successfully created as 'parks_final_report.xlsx'!")
```
